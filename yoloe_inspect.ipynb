## YOLOE: Real-Time Seeing Anything

YOLOE is a zero-shot, promptable YOLO model designed for **open-vocabulary object detection and segmentation**. Unlike traditional YOLO models that rely on fixed class labels, YOLOE allows detection of any object using **text prompts**, **visual prompts**, or **internal vocabulary**—all in real time.

### 🧠 Architecture Overview
YOLOE retains the typical YOLO structure:
- **Backbone:** CSP-Darknet (for feature extraction)
- **Neck:** PAN-FPN (for multi-scale feature fusion)
- **Head:** Anchor-free, decoupled detection (predicts objectness, class, and box independently)

### 🔍 Key Innovations
1. **RepRTA (Re-parameterizable Region-Text Alignment):**  
   Refines CLIP-based text embeddings for text-prompted detection. Folded into the main model at inference = zero overhead.

2. **SAVPE (Semantic-Activated Visual Prompt Encoder):**  
   Encodes reference images to enable **visual-prompted detection**—useful for one-shot tasks like logo detection.

3. **LRPC (Lazy Region-Prompt Contrast):**  
   Enables **prompt-free open-set recognition** by comparing internal embeddings against large vocabularies (e.g., LVIS, Objects365).

````python
from ultralytics import YOLOE
import torch

# Initialize a YOLOE model
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
model = YOLOE("yoloe-11l-seg.pt").to(device)


# Set text prompt to detect person and bus. You only need to do this once after you load the model.
names = ["black dog", "white dog"]
model.set_classes(names, model.get_text_pe(names))

# Run detection on the given image
results = model.predict("images/dogs.mp4", show=True, save=True)

# Show results
results[0].show()
````

## Prompt-Free YOLOE Models

YOLOE also includes **prompt-free variants** that come with a built-in vocabulary. These models don't require any prompts and work similarly to traditional YOLO models. Instead of relying on user-provided labels or visual examples, they detect objects from a predefined list of **4,585 classes**, based on the tag set used by the Recognize Anything Model Plus (RAM++).

🔗 [View predefined class list (RAM++ tag set)](https://github.com/xinyu1205/recognize-anything/blob/main/ram/data/ram_tag_list.txt)
